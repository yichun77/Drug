```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
plt.rcParams['font.family'] = 'Microsoft YaHei'
plt.rcParams['axes.unicode_minus'] = False
```

```python
df = pd.read_excel(r"C:\Users\adayc\Desktop\code\Drug\DA_data\filtered_DA_daily_meal.xlsx")  
print(df.columns)
```

```python
base_columns = ['Sampleid', 'DAY', 'TIME']
feature_columns = [col for col in df.columns if col not in base_columns]
sub_tables = {}
for col in feature_columns:
    sub_df = df[base_columns + [col]].copy()
    sub_tables[col] = sub_df
```

```python
output_dir = r"C:\Users\adayc\Desktop\code\Drug\reports\results\figures\feature_meal"

for col, sub_df in sub_tables.items():
    sub_df['TIME'] = sub_df['TIME'].astype(str)

    fig, axes = plt.subplots(4, 1, figsize=(14, 16), sharex=False)
    fig.suptitle(f"{col} 在不同时段的分布（按 Sampleid）", fontsize=18)

    for i, time_point in enumerate(['1', '2', '3', '4']):
        ax = axes[i]
        subset = sub_df[sub_df['TIME'] == time_point]

        if not subset.empty:
            sns.boxplot(
                data=subset,
                x='Sampleid',
                y=col,
                ax=ax,
                palette='Set2'
            )
            ax.set_title(f'TIME = {time_point}', fontsize=14)
            ax.set_xlabel('Sampleid', fontsize=12)
            ax.set_ylabel(col, fontsize=12)
            ax.tick_params(axis='x', rotation=45)  
        else:
            ax.set_title(f'TIME = {time_point}（无数据）')
            ax.axis('off')

    plt.tight_layout(rect=[0, 0, 1, 0.96])
    safe_col = col.replace("/", "_")
    filename = os.path.join(output_dir, f"{safe_col}.png")
    plt.savefig(filename)
    plt.close()
```
